# What are Transformers?

Transformers are Deep Learning models introduced by Google in 2017. They are designed to process and understand natural language efficiently.

Unlike traditional NLP models, Transformers can understand the context and relationships between words in a sentence.

Examples of applications:
- ChatGPT
- Google Translate
- AI Chatbots
- Text Summarization
- Question Answering
- Language Translation

# What is Transfer Learning?

Transfer Learning is a technique where a model that has already been trained on a large amount of data is reused for a new task.

Instead of training a model from scratch, we use a pre-trained model and fine-tune it or use it directly for prediction.

Benefits:
- Saves time
- Requires less data
- Provides better accuracy
- Faster development

# What is Hugging Face?

Hugging Face is an open-source platform that provides thousands of pre-trained Artificial Intelligence models.

It allows developers to use state-of-the-art NLP models with just a few lines of Python code.

Popular Tasks:
- Sentiment Analysis
- Text Classification
- Question Answering
- Text Summarization
- Text Generation
- Translation
- Named Entity Recognition (NER)

# What is BERT?

BERT stands for Bidirectional Encoder Representations from Transformers.

It was developed by Google.

BERT reads text in both directions (left-to-right and right-to-left), allowing it to understand the context of words more effectively.

Applications:
- Question Answering
- Sentiment Analysis
- Text Classification
- Named Entity Recognition

# What is GPT?

GPT stands for Generative Pre-trained Transformer.

It was developed by OpenAI.

GPT is designed to generate human-like text based on the input provided by the user.

Applications:
- Chatbots
- Content Writing
- Story Generation
- Code Generation
- Text Completion

```python
from transformers import pipeline

# Load Sentiment Analysis Pipeline
sentiment = pipeline("sentiment-analysis")

# Test Sentence
result = sentiment("Hugging Face predicts sentiment.")

print(result)
```

```python
from transformers import pipeline

sentiment = pipeline(
    "sentiment-analysis",
    model="distilbert/distilbert-base-uncased-finetuned-sst-2-english"
)

result = sentiment("This movie is amazing.")

print(result)
```

## Observation

The Hugging Face model correctly identified the sentiment of the sentence as Negative.

```python
from transformers import pipeline

qa = pipeline("question-answering")

context = """
Python is a popular programming language used in Artificial Intelligence,
Machine Learning, Data Science and Web Development.
"""

question = "What is Python used for?"

result = qa(question=question, context=context)

print(result)
```
