```python
import pandas as pd
```

```python
data = pd.read_csv('loan_check.csv')
```

```python
data.head()
```

```python
data.drop(columns = ['loan_id'], inplace = True)
```

```python
data.columns
```

```python
data.columns= data.columns.str.strip()  # remove extra space in columns 
```

```python
data.columns
```

```python
data['assets'] = data.residential_assets_value+data.commercial_assets_value+data.luxury_assets_value+data.bank_asset_value	
```

```python
# all types of assets columns combine as assets
```

```python
# drop self_employed column and all types of assets columns as we have combined them as assets column
data.drop(columns=['residential_assets_value','bank_asset_value','commercial_assets_value','luxury_assets_value','self_employed'],axis = 1,inplace=True)

```

```python
data.head()
```

```python
data.isnull().sum()
```

```python
data.education.unique()  # categorical values name 
```

```python
def clean_data(st):
    st = st.strip()
    return st
```

```python
clean_data(' Graduate')
```

```python
data.education = data.education.apply(clean_data)
```

```python
data.education.unique()
```

```python
data['education'] = data['education'].replace(['Graduate', 'Not Graduate'],[1,0])
```

```python
data.head()
```

```python
data.loan_status.unique()
```

```python
data.loan_status = data.loan_status.apply(clean_data)
```

```python
data.loan_status = data.loan_status.replace(['Approved', 'Rejected'],[1,0])
```

```python
data.head()   # completete data as clean 
```

```python
from sklearn.model_selection import train_test_split
```

```python
input_data = data.drop(columns=['loan_status'])
output_data = data['loan_status']
```

```python
output_data
```

```python
input_data
```

```python
x_train,x_test,y_train,y_test = train_test_split(input_data,output_data,test_size=0.2,random_state=42) # splitting 
```

```python
x_train.shape,x_test.shape,y_train.shape,y_test.shape
```

```python
from sklearn.preprocessing import StandardScaler  # for improving the performances of model
```

```python
scaler = StandardScaler()
```

```python
x_train_scaled = scaler.fit_transform(x_train)
```

```python
x_test_scaled = scaler.transform(x_test)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression()
```

```python
model.fit(x_train_scaled,y_train)
```

```python
model.score(x_test_scaled,y_test)
```

```python
pred_data = pd.DataFrame([['2',	'1',	'9600000',	'29900000',	'12',	'778',	'50700000']],columns = ['no_of_dependents',	'education',	'income_annum',	'loan_amount',	'loan_term',	'cibil_score',	'assets'])
```

```python
pred_data =scaler.transform(pred_data)
```

```python
model.predict(pred_data)
```

```python
# '1' shows means approved 
```

```python
import pickle as pk
```

```python
from sklearn.tree import DecisionTreeClassifier

dt_model = DecisionTreeClassifier(random_state=42)

dt_model.fit(x_train_scaled,y_train)

```

```python
from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier(random_state=42)

rf_model.fit(x_train_scaled,y_train)

```

```python
from sklearn.neighbors import KNeighborsClassifier

knn_model = KNeighborsClassifier(n_neighbors=5)

knn_model.fit(x_train_scaled, y_train)

```

```python
from sklearn.metrics import accuracy_score
from sklearn.metrics import precision_score
from sklearn.metrics import recall_score
from sklearn.metrics import f1_score
from sklearn.metrics import roc_auc_score
```

```python
# logistic regression model evaluation
y_pred_lr = model.predict(x_test_scaled)
print("Logistic Regression Metrics:")
print("Accuracy :", accuracy_score(y_test, y_pred_lr))
print("Precision:", precision_score(y_test, y_pred_lr))
print("Recall   :", recall_score(y_test, y_pred_lr))
print("F1 Score :", f1_score(y_test, y_pred_lr))
print("ROC AUC  :", roc_auc_score(y_test, y_pred_lr))
```

```python
# decision tree model evaluation
y_pred_dt = dt_model.predict(x_test_scaled)
print("Decision Tree Metrics:")
print("Accuracy :", accuracy_score(y_test, y_pred_dt))
print("Precision:", precision_score(y_test, y_pred_dt))
print("Recall   :", recall_score(y_test, y_pred_dt))
print("F1 Score :", f1_score(y_test, y_pred_dt))
print("ROC AUC  :", roc_auc_score(y_test, y_pred_dt))
```

```python
# Random Forest model evaluation
y_pred_rf = rf_model.predict(x_test_scaled)
print("Random Forest Metrics:")
print("Accuracy :", accuracy_score(y_test, y_pred_rf))
print("Precision:", precision_score(y_test, y_pred_rf))
print("Recall   :", recall_score(y_test, y_pred_rf))
print("F1 Score :", f1_score(y_test, y_pred_rf))
print("ROC AUC  :", roc_auc_score(y_test, y_pred_rf))
```

```python
# knn model evaluation
y_pred_knn = knn_model.predict(x_test_scaled)
print("KNN Metrics:")
print("Accuracy :", accuracy_score(y_test, y_pred_knn))
print("Precision:", precision_score(y_test, y_pred_knn))
print("Recall   :", recall_score(y_test, y_pred_knn))
print("F1 Score :", f1_score(y_test, y_pred_knn))
print("ROC AUC  :", roc_auc_score(y_test, y_pred_knn))
```
