Natural Language Processing (NLP) is a branch of Artificial Intelligence that helps computers understand, process, and analyze human language.

NLP is used in many real-world applications such as chatbots, language translation, spam detection, sentiment analysis, voice assistants, and text classification.

In this notebook, we will learn basic NLP preprocessing techniques such as Tokenization, Stopword Removal, Stemming, Lemmatization, TF-IDF, and Sentiment Analysis.

```python
# Import Libraries

import pandas as pd
import nltk

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer

from sklearn.feature_extraction.text import TfidfVectorizer
```

## Sample Sentences

The following five movie review sentences will be used to demonstrate different NLP preprocessing techniques.

```python
sentences = [
    "The movie was absolutely amazing and enjoyable.",
    "I did not like the acting in the movie.",
    "The story was boring but the music was good.",
    "This is one of the best films I have ever watched.",
    "The movie was too long and disappointing."
]

for i, sentence in enumerate(sentences, start=1):
    print(f"Sentence {i}: {sentence}")
```

# Tokenization

Tokenization is the process of breaking a sentence into individual words called tokens.

It is the first step in Natural Language Processing.

```python
for i, sentence in enumerate(sentences, start=1):

    tokens = word_tokenize(sentence)

    print(f"\nSentence {i}")
    print("Original :", sentence)
    print("Tokens   :", tokens)
```

### Observation

Tokenization divides a sentence into individual words, making it easier for computers to understand and process text.

# Stopword Removal

Stopwords are common English words such as "the", "is", "was", "and", "in".

Removing stopwords helps reduce unnecessary words and improves text processing.

```python
stop_words = set(stopwords.words("english"))

for i, sentence in enumerate(sentences, start=1):

    tokens = word_tokenize(sentence)

    filtered = []

    for word in tokens:
        if word.lower() not in stop_words:
            filtered.append(word)

    print(f"\nSentence {i}")
    print("Original :", sentence)
    print("Filtered :", filtered)
```

### Observation

Stopword Removal removes common words that usually do not add much meaning to a sentence.

# Stemming

Stemming reduces words to their root form by removing prefixes or suffixes.

The resulting word may not always be a proper English word.

```python
stemmer = PorterStemmer()

for i, sentence in enumerate(sentences, start=1):

    tokens = word_tokenize(sentence)

    stemmed = []

    for word in tokens:
        if word.isalpha():
            stemmed.append(stemmer.stem(word))

    print(f"\nSentence {i}")
    print("Original :", sentence)
    print("Stemmed  :", stemmed)
```

### Observation

Stemming converts words into their root form.

Example:

playing → play

enjoyable → enjoy

amazing → amaz

# Lemmatization

Lemmatization converts words into meaningful dictionary words.

It produces better results than stemming.

```python
lemmatizer = WordNetLemmatizer()

for i, sentence in enumerate(sentences, start=1):

    tokens = word_tokenize(sentence)

    lemmas = []

    for word in tokens:
        if word.isalpha():
            lemmas.append(lemmatizer.lemmatize(word))

    print(f"\nSentence {i}")
    print("Original      :", sentence)
    print("Lemmatization :", lemmas)
```

### Observation

Lemmatization converts words into meaningful base forms.

It is more accurate than stemming because the output is usually a valid English word.

# TF-IDF

TF-IDF (Term Frequency - Inverse Document Frequency) converts text into numerical values.

Machine Learning algorithms cannot understand text directly.

TF-IDF assigns higher importance to meaningful words and lower importance to common words.

```python
vectorizer = TfidfVectorizer()

X = vectorizer.fit_transform(sentences)

feature_names = vectorizer.get_feature_names_out()

tfidf = pd.DataFrame(
    X.toarray(),
    columns=feature_names
)

tfidf
```

### Observation

TF-IDF converts text into numbers.

These numerical values are used by Machine Learning algorithms for text classification.

# Sentiment Analysis

Sentiment Analysis is the process of identifying whether a piece of text expresses a Positive, Negative, or Neutral opinion.

Example:

"I loved this movie." → Positive

"The movie was boring." → Negative

```python
reviews = [
    "Movie was amazing",
    "Excellent acting",
    "Worst movie ever",
    "I loved this movie",
    "Waste of time"
]

sentiments = [
    "Positive",
    "Positive",
    "Negative",
    "Positive",
    "Negative"
]

df = pd.DataFrame({
    "Review": reviews,
    "Sentiment": sentiments
})

df
```
